## WHAT IS A JUPYTER NOTEBOOK?

A Jupyter Notebook is an interactive computing environment that lets you write and run code in small chunks (called "cells"), while also mixing in text, equations, and visualizations. It is widely used for data analysis, machine learning, teaching, and experimentation.

Instead of writing a full script and running it all at once, you can execute code step by step and immediately see results.

MAIN PARTS OF A JUPYTER NOTEBOOK

1) NOTEBOOK FILE (.ipynb)
   - A file that stores everything: code, text, outputs, and metadata.
   - It is JSON under the hood, but you interact with it visually.

2) CELLS
   Can be executable cells or markdown cells

3) OUTPUT AREA

4) KERNEL (VERY IMPORTANT)
   - The "engine" that runs your code.
   - Keeps track of variables and state.

   Key ideas about the kernel:
   - It executes code in cells.
   - It remembers variables between cells.
   - It can be restarted (which clears memory).
   - Different kernels support different languages (Python, R, Julia, etc.).


5) NOTEBOOK INTERFACE



COMPARISON: JUPYTER NOTEBOOKS VS OTHER DEVELOPMENT APPROACHES

| Feature                    | Jupyter Notebook                     | VS Code / IDE                     | Code in VM / Scripts              |
|--------------------------|-------------------------------------|----------------------------------|----------------------------------|
| Execution Style          | Cell-by-cell                        | File-based (run whole or debug)  | File-based                       |
| Interactivity            | Very high                           | Medium                           | Low                              |
| Ease for Beginners       | Very easy                           | Moderate                         | Harder                           |
| Debugging                | Basic                               | Advanced tools                   | Depends on setup                 |
| Reproducibility          | Can be messy if cells run out of order | Better structured               | Very structured                  |
| State Management         | Persistent (via kernel)             | Explicit (variables reset on run)| Explicit                         |
| Best For                 | Data analysis, teaching, prototyping| Software development             | Production systems               |
| Visualization            | Built-in and immediate              | Requires setup                   | Requires setup                   |
| Performance              | Not ideal for large systems         | Better                           | Best (controlled environments)   |
| Environment Isolation    | Limited unless configured           | Good (virtual envs, containers)  | Strong (VM-level isolation)      |


KEY LIMITATIONS OF JUPYTER NOTEBOOKS

- Execution order can become confusing
- Harder to maintain large codebases
- Not ideal for production applications **
- Version control (e.g., Git) is less clean than plain code files

## What is a Module / Library?

In Python, a **module** is a single file that contains reusable code (functions, variables, classes).

A **library** (or package) is a collection of modules grouped together to provide more advanced functionality.

**Hierarchy:**
- Module → single `.py` file
- Library / Package → folder containing multiple modules

**Examples:**
- `math` → built-in module
- `numpy` → library for numerical computing
- `pandas` → library for data analysis

# Requirements.txt

A `requirements.txt` file is a simple text file that lists all the Python libraries (and their versions) that a project depends on.

It is used to **recreate the exact same environment** on another machine or at a later time.

---

__Example:__

```txt
pandas==1.5.3
numpy==1.24.0
matplotlib==3.7.1
```

__Why It’s Important:__

Reproducibility → same environment everywhere
Collaboration → teammates get identical setup
Deployment → servers run the same dependencies
Debugging → avoids “works on my machine” issues

````python
import sys

!{sys.executable} -m pip freeze > requirements.txt
````

# Virtual enviroments

Virtual environments create **isolated Python environments per project**, so dependencies don’t interfere with each other.

---

## Problem

Different projects may need different library versions:

- Project A → `pandas==1.5.3`
- Project B → `pandas==2.2.0`

Without isolation → conflict  
With virtual envs → no problem

---

## Key Benefits

- **Isolation**  
  Each project has its own libraries and versions

- **Reproducibility**  
  Share exact setup:
  ```bash
  pip freeze > requirements.txt
