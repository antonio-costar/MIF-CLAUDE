## Why Packaging a Python Library is Useful

Packaging a Python library turns a collection of scripts into a reusable, installable, and distributable software component.

__1. Reusability__
Packaging allows your code to be reused across projects without duplication, enabling consistent and modular development.

__2. Version Control & Releases__
It enables structured versioning (e.g., v1.0.0), making updates traceable, reproducible, and safe to deploy.

__3. Shareability__
A packaged library can be easily shared and installed by others, removing the need for manual file copying or setup.

__4. Operationalization__
Packaging transforms code into production-ready software that can be integrated into pipelines, deployed systems, and automated workflows.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def plot_cat2violin(df, categorical_col, numerical_col):
    # Calculate the mean of the numerical column for each category in the categorical column
    category_means = df.groupby(categorical_col)[numerical_col].mean().sort_values(ascending=False)
    
    # Create a new column with ordered categories
    df['sorted_category'] = pd.Categorical(df[categorical_col], categories=category_means.index, ordered=True)
    
    plt.figure(figsize=(10, 6))
    
    sns.violinplot(x='sorted_category', y=numerical_col, data=df)
    
    # Add a line connecting the mean points
    plt.plot(category_means.index, category_means.values, color='r', marker='o', linestyle='-')
    
    # Add mean values at the bottom, just above the x-axis
    for i, mean in enumerate(category_means.values):
        plt.text(i, df[numerical_col].min() - 1, f'{mean:.2f}', color='r', ha='center', va='top')
    
    plt.title(f'{numerical_col} Distribution by {categorical_col} (Sorted)')
    plt.xlabel(categorical_col)
    plt.ylabel(numerical_col)
    
    plt.xticks(rotation=45)
    plt.tight_layout()
    
    plt.show()

    del df['sorted_category']

    print('source: Notebook')
```

```python
plot_cat2violin(df, categorical_col = 'occupation'
                                        , numerical_col= 'educationnum')
```

## Git interaction


Git is a distributed version control system used to track changes in code, coordinate work among developers, and maintain a complete history of a project.

### Why Git is Useful

Git allows developers to:
- Track changes over time and revert to previous versions if needed
- Work in parallel using branches without interfering with the main codebase
- Collaborate efficiently across teams and locations
- Maintain a reliable history of every modification
- Safely experiment with new features without breaking production code

In modern software development, Git is essential for collaboration, reproducibility, and project organization.

---

### Summary of Core Git Actions

| Action  | Description |
|----------|-------------|
| init     | Initializes a new Git repository in a project directory |
| clone    | Creates a local copy of a remote repository |
| add      | Stages changes (files) for a commit |
| commit   | Saves a snapshot of staged changes to the repository history |
| status   | Shows the current state of the working directory and staging area |
| branch   | Creates or lists branches for parallel development |
| checkout | Switches between branches or commits |
| merge    | Combines changes from one branch into another |
| pull     | Fetches and integrates changes from a remote repository |
| push     | Sends local commits to a remote repository |

---

### Summary
Git enables structured, safe, and collaborative software development by tracking changes, managing versions, and supporting parallel workflows through branching."

## Create a Git Repo


We first locate our local repo. We assume we already have it. But later on we will see how to clone an external repo. This first example shows you how to share your development to the project:

```python
import os

from pathlib import Path

# CHANGE YOUR DIRECTORY TO THE LOCAL PATH OF YOUR PROJECT
local_path = Path(r"C:/Users/antoniocosta/Desktop/MIF/my_toolbox")
os.chdir(local_path)
```

```python
# create repository using GitHub API
import requests

GITHUB_USERNAME = "antonio_costar"
REPO_NAME = "my_toolbox"
TOKEN = "REDACTED_TOKEN"  # Personal access token (keep this private)

# GitHub API endpoint to create a new repository
url = "https://api.github.com/user/repos"

# Authentication and request headers
headers = {
    "Authorization": f"token {REDACTED_TOKEN}",
    "Accept": "application/vnd.github+json"
}

# Repository configuration payload
data = {
    "name": REPO_NAME,
    "private": False,  # set True for private repo
    "description": "Custom visualization toolbox",
    "auto_init": False  # do not initialize with README
}

# Send request to GitHub API to create repo
response = requests.post(url, json=data, headers=headers)

# Print result of the request
print("Status:", response.status_code)
print(response.json())
```

```python
# Import subprocess module to allow execution of system commands (e.g., git)
import subprocess

# Helper function to run shell commands safely from Python
def run(cmd):
    # Execute the command in a shell and capture output + errors
    result = subprocess.run(
        cmd,
        shell=True,
        capture_output=True,
        text=True
    )

    # Print standard output from the command (normal results)
    print(result.stdout)

    # Print standard error output (warnings or errors)
    if result.stderr:
        print(result.stderr)

    # If the command failed (non-zero exit code), raise an exception
    if result.returncode != 0:
        raise Exception(result.stderr)
```

```python
# Initialize repo
run("git init")

# Add all files
run("git add .")
```

```python
import subprocess

# Construct the remote repository URL using GitHub username and repository name
remote_url = f"https://github.com/{GITHUB_USERNAME}/{REPO_NAME}.git"

# check if origin exists
result = subprocess.run(
    "git remote",
    shell=True,
    capture_output=True,
    text=True
)

remotes = result.stdout.split()

if "origin" in remotes:
    run(f"git remote set-url origin {remote_url}")
else:
    run(f"git remote add origin {remote_url}")
```

```python
# Show the current state of the working directory:
# - which branch you're on
# - staged/unstaged changes
# - untracked files
run("git status")

# Display the configured remote repositories:
# - shows URLs for fetch and push (e.g., origin)
run("git remote -v")

# Show the latest commit history in a compact format:
# - -n 3 limits output to the last 3 commits
run("git log --oneline -n 3")
```

## Contributing to a project
Now, lets say you want to contribute to an existing repo. This is, by uploading your changes.

<img src="images//git_workflow.png" width="600" img style="float: left;">

```python
# -------------------------------
# Feature branch workflow
# -------------------------------

# Create and switch to a new feature branch
# This isolates development work from the main branch
run("git checkout -b feature-update")

# Stage all modified, new, and deleted files
# Prepares changes to be included in the next commit
run("git add -A")

# Create a commit (snapshot of current changes)
# This saves the state of the project with a descriptive message
run('git commit -m "My feature update"')

# Push the feature branch to the remote repository (GitHub)
# -u sets upstream tracking so future pushes can be simplified
run("git push -u origin feature-update")


# (Optional) Push main branch - NOT needed in feature workflow unless main has changes
# run("git push origin main")


# Check current repository status
# Useful to confirm working directory is clean and no pending changes exist
run("git status")
```
