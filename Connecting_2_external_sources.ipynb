```python
data = []  # list to store each product's extracted information

for p in products:  # loop through each product element in the HTML

    # extract product name from <a class="title"> tag and clean whitespace
    name = p.select_one("a.title").get_text(strip=True)

    # extract price from <span itemprop="price"> and convert to float
    price = p.select_one('span[itemprop="price"]').get_text(strip=True)
    price = float(price.replace("$", ""))  # remove dollar sign before conversion

    # extract number of reviews; if missing, default to 0
    reviews_tag = p.select_one('span[itemprop="reviewCount"]')
    reviews = int(reviews_tag.get_text(strip=True)) if reviews_tag else 0

    # extract rating from data attribute; if missing, set to None
    rating_tag = p.select_one("p[data-rating]")
    rating = float(rating_tag["data-rating"]) if rating_tag else None

    # store cleaned/extracted data in dictionary and add to list
    data.append({
        "name": name,
        "price": price,
        "number_of_reviews": reviews,
        "average_rating": rating
    })

# convert list of dictionaries into a pandas DataFrame
df = pd.DataFrame(data)

df
```

```python
import yfinance as yf
import pandas as pd

# Pick any ticker, e.g. "AAPL", "MSFT", "TSLA"
ticker = "AAPL"

# Get 1 day of data at intraday intervals (similar to the "1D" view in your screenshot)
data = yf.download(ticker, period="1d", interval="5m")

# Build the final dataframe with just date and price
df = data.reset_index()[["Datetime", "Close"]]
df.columns = ["date", "price"]

print(df.head())
```
