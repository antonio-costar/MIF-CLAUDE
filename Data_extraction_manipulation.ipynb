__Popular transformation__

## One Hot Encoding transformation

```python
df.groupby(['occupation']).size().reset_index().rename(columns={0: 'count'}).sort_values(by='count', ascending=False)
```

__Example: One-Hot Encoding of "occupation" column__

After one-hot encoding:

Each occupation becomes its own binary column (0 = not present, 1 = present)

-----------------------------------------------------------------------------------------
| Prof-specialty | Craft-repair | Exec-managerial | Adm-clerical | Sales |
|----------------|--------------|------------------|--------------|-------|
| 1              | 0            | 0                | 0            | 0     |  <- Prof-specialty
| 0              | 1            | 0                | 0            | 0     |  <- Craft-repair
| 0              | 0            | 1                | 0            | 0     |  <- Exec-managerial
| 0              | 0            | 0                | 1            | 0     |  <- Adm-clerical
| 0              | 0            | 0                | 0            | 1     |  <- Sales
-----------------------------------------------------------------------------------------

Key idea:
- Only ONE column is 1 per row (if using standard one-hot encoding)
- All other columns are 0
- This converts categorical data into numeric format (useful for ML models)

```python
import pandas as pd
import numpy as np

customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4, 5, 6],
    "name": ["Alice", "Bob", "Charlie", "Diana", "Eve", "Frank"],
    "city": ["London", "Paris", "Berlin", "Madrid", "Rome", "Lisbon"]
})

agents = pd.DataFrame({
    "agent_id": [10, 20],
    "agent_name": ["Agent Smith", "Agent Johnson"],
    "region": ["North", "South"]
})

orders = pd.DataFrame({
    "order_id": range(1001, 1021),
    "customer_id": [
        1, 1, 2, 2, 3, 3, 3, 4, 4, 5,
        5, 5, 6, 6, 6, 1, 2, 7, 8, 3   # 7 and 8 don't exist → useful for teaching
    ],
    "agent_id": [
        10, 20, 10, 10, 20, 20, 10, 10, 20, 10,
        20, 20, 10, 10, 20, 20, 10, 10, 30, 20  # 30 doesn't exist → teaching case
    ],
    "amount": [
        120, 80, 50, 60, 200, 150, 90, 40, 75, 300,
        220, 180, 130, 170, 160, 110, 95, 210, 400, 85
    ]
})
```

```python
# what orders where performed by which customers?
orders_customers = pd.merge(
    orders,
    customers,
    on="customer_id",
    how="left"
)

orders_customers
```
