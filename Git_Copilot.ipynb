````python
def integrate_credit_default(credit, default):
    """
    Integrates credit and default data into a single structure.
    
    Parameters:
    credit (list): A list of credit amounts.
    default (list): A list indicating default status (0 for no default, 1 for default).
    
    Returns:
    list: A list of tuples where each tuple contains a credit amount and its corresponding default status.
    """
    return list(zip(credit, default))
````
