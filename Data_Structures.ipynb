## What is a data structure?

A **data structure** is a way of organizing and storing data so that it can be accessed, modified, and processed efficiently. Different data structures are designed for different types of data and operations—for example, storing an ordered collection, ensuring values are unique, or associating a key with a value. Choosing the right data structure can make a program simpler, faster, and more memory-efficient.

## Common Python data structures and data types

| Data type | Example | Benefits | Cons | Appropriate databases / storage |
|---|---|---|---|---|
| **String (`str`)** | `"Hello"` | Simple way to represent text; easy to manipulate and search | Cannot be modified in place; can use significant memory for large text | SQL `VARCHAR` / `TEXT`, NoSQL document fields |
| **Integer (`int`)** | `42` | Exact whole-number representation; efficient for counting and IDs | Cannot represent decimals; Python integers can use more memory for very large values | SQL `INTEGER` / `BIGINT`, NoSQL numeric fields |
| **Float (`float`)** | `3.14` | Useful for measurements and approximate decimal calculations | Floating-point precision can cause unexpected results | SQL `FLOAT` / `DOUBLE`, NoSQL numeric fields |
| **Boolean (`bool`)** | `True` | Clear representation of two states: true/false | Only represents two possible states | SQL `BOOLEAN`, NoSQL boolean fields |
| **Set (`set`)** | `{1, 2, 3}` | Very fast membership checks; automatically removes duplicates; useful for set operations | No indexing; duplicates are not preserved | SQL tables with `UNIQUE` constraints, NoSQL arrays with application-level uniqueness |
| **None (`NoneType`)** | `None` | Represents the absence of a value | Does not contain useful data itself; must be handled explic
`null` |
| **List (`list`)** | `[10, 20, 30]` | Flexible; allows duplicates; easy to add/remove items; supports indexing | Searching for an item can be slow for large lists | SQL table with multiple rows, NoSQL arrays |
| **Tuple (`tuple`)** | `(10, 20, 30)` | Immutable; useful for fixed collections; can be used as dictionary keys | Cannot be changed after creation | SQL table/row, NoSQL arrays or embedded documents |
| **Dictionary (`dict`)** | `{"name": "Alice", "age": 25}` | Fast lookup using keys; naturally represents key-value data | Keys must be unique; can become complex when deeply nested | SQL tables with columns, NoSQL document databases such as MongoDB |
| **Array (`array.array`)** | `array('i', [1, 2, 3])` | Stores elements of the same basic type efficiently; uses less memory than a Python list for numerical data | Less flexible than a list; fewer operations; not as powerful for numerical computing as NumPy | SQL numeric columns, binary files, specialized data storage |
| **NumPy array (`numpy.ndarray`)** | `np.array([1, 2, 3])` | Very efficient numerical operations; supports multidimensional data; vectorization; mathematical and scientific computing | Requires NumPy; elements generally have the same data type; can use significant memory for very large arrays | SQL tables, scientific data files, HDF5, Parquet, cloud/object storage |
| **Pandas DataFrame** | `pd.DataFrame({"name": ["Ana", "Bob"], "age": [25, 30]})` | Excellent for tabular data; labeled rows and columns; powerful filtering, grouping, aggregation, and data cleaning | Higher memory usage; can be slower than NumPy for pure numerical operations; requires pandas | SQL tables, CSV, Excel, Parquet, databases, dat → use `None`

# Floating-Point Types: `float32`, `float64`


| Type | Bits | Bytes | Sign | Exponent | Significand | Approx. decimal precision |
|---|---:|---:|---:|---:|---:|---:|
| `float32` | 32 | 4 | 1 bit | 8 bits | 23 bits | ~7 digits |
| `float64` | 64 | 8 | 1 bit | 11 bits | 52 bits | ~15–16 digits |

__Sign__ → determines whether the number is positive or negative.  
__Exponent__ → determines the scale of the number.  
__Significand__ → determines the precision of the number.

<div style="font-size: 2em;">

$$
(-1)^{\text{sign}} \times \text{significand} \times 2^{\text{exponent}}
$$

</div>

further readings: https://www.mathsisfun.com/binary-number-system.html

````python
# a+b
print(0.1 + 0.2)
````

The code performs 0.1 + 0.2 using both float32 and float64, which store the numbers with different levels of precision.
The first print shows Python's rounded, human-friendly representation, while format(..., "20f") exposes the stored approximation with207 decimal places.
This demonstrates that float64 is more precise than float32, but neither can represent 0.1 and 0.2 exactly in binary.

````python
import numpy as np

# float32
a32 = np.float32(0.1)
b32 = np.float32(0.2)

print("float32:")
print(a32 + b32)
print(format(a32 + b32, ".20"))


# float64
a64 = np.float64(0.1)
b64 = np.float64(0.2)

print("\nfloat64:")
print(a64 + b64)
print(format(a64 + b64, ".20"))
````

# Common structures
__This example shows that different Python data structures can store the same numerical values using very different amounts of memory:__

- A **Python list** stores references to individual Python objects, so it requires considerably more memory.
- A **Python `array`** stores values directly using a fixed data type, making it much more memory-efficient.
- `float32` uses **4 bytes per value**, while `float64` uses **8 bytes per value**, so `float64` requires twice as much memory.
- This difference becomes important when a finance department processes millions or billions of transactions, prices, balances, or other numerical values: even a small amount of extra memory per value can become a significant amount of storage.
- However, **memory efficiency should not be considered independently of precision**. Financial applications often require exact decimal calculations, so `float32` should not simply be chosen because it uses less memory.

````python
import sys
import numpy as np
from array import array

values = [0.5, 0.25, 0.75, 1.0]

# Python list
python_list = values

# Python array: 64-bit floating point
array_float64 = array('d', values)

# Python array: 32-bit floating point
array_float32 = array('f', values)

# Display the values and memory occupied by the data
print("Python list:")
print(python_list)
print(f"Approx. memory for elements: {len(python_list) * sys.getsizeof(python_list[0])} bytes")
print(f"List container itself: {sys.getsizeof(python_list)} bytes")

print("\nPython array - float32:")
print(array_float32)
print(f"Memory: {array_float32.itemsize * len(array_float32)} bytes")

print("\nPython array - float64:")
print(array_float64)
print(f"Memory: {array_float64.itemsize * len(array_float64)} bytes")
````

## Common structures side by side
| Operation              | list              | array.array        | dict                    |
|------------------------|------------------|--------------------|-------------------------|
| Access by index        | O(1)             | O(1)               | N/A                     |
| Search by value        | O(n)             | O(n)               | O(n)                    |
| Lookup by key          | N/A              | N/A                | O(1)             |
| Append (add at end)    | O(1)   | O(1)               | O(1)                    |
| Pop last               | O(1)             | O(1)               | O(1)                    |
| Pop first              | O(n)             | O(n)               | N/A                     |
| Insert in middle       | O(n)             | O(n)               | N/A                     |
| Delete by key/index    | O(n) (by value)  | O(n)               | O(1)                    |
| Memory efficiency      | Medium           | High               | Low                     |
| Allows mixed types     | Yes              | No                 | Yes (values)            |
| Maintains order        | Yes              | Yes                | Yes (Python 3.7+)       |

````python
from array import array

# List
numbers_list = [3, 1, 4]

# Array (only one type allowed, here integers 'i')
numbers_array = array('i', [3, 1, 4])

# Dictionary (key → value)
numbers_dict = {"a": 3, "b": 1, "c": 4}
````

__Adding elements__

````python
# LIST
numbers_list.append(2)
print(numbers_list)  

# ARRAY
numbers_array.append(2)
print(numbers_array)  

# DICTIONARY
numbers_dict["d"] = 2
print(numbers_dict)  
````

__Removing elements__

````python
# LIST (removes last by default)
value = numbers_list.pop()
print(value)        
print(numbers_list) 

# ARRAY
value = numbers_array.pop()
print(value)          
print(numbers_array)  

# DICTIONARY
value = numbers_dict.pop("d")
print(value)          
print(numbers_dict) 
````

__Soriting elements__

````python
# LIST
numbers_list.sort(reverse=False)
print(numbers_list) 

# ARRAY
numbers_array = array('i', sorted(numbers_array))
print(numbers_array)

# DICTIONARY (sort by values)
sorted_items = sorted(numbers_dict.items(), key=lambda x: x[1])
print(sorted_items)
````

__comment:__ `reverse`= `desc` (SQL)

__Iterating elements__

````python
# LIST
for x in numbers_list:
    print(x)
print('_________')
# ARRAY
for x in numbers_array:
    print(x)
print('_________')
# DICTIONARY
for key, value in numbers_dict.items():
    print(key, value)
````

## Lists vs Dictionaries — Quick Summary

### List (sequence)
- Stores elements in order
- Access by index → O(1)
- Search by value → O(n) (slow, scans elements)

Example:
    clients = ["Alice", "Bob", "Carlos"]

    clients[0]        # "Alice" (fast)
    "Carlos" in clients  # checks one by one (slower)

---

### Dictionary (key → value)
- Stores data as key-value pairs
- Access by key → O(1)
- No need to search

Example:
    clients = {"Alice": 1000, "Bob": 1500, "Carlos": 800}

    clients["Carlos"]  # 800 (direct access)

---

### Core Difference
- List → position-based → must search → time grows with size  
- Dict → key-based → direct lookup → constant time  

---

### When to use
- Use list → ordered data, indexing  
- Use dict → fast lookup, labeled data  

````python
clients_list = [
    ["Alice", 1000],
    ["Bob", 1500],
    ["Carlos", 800]
]

clients_dict = {
    "Alice": 1000,
    "Bob": 1500,
    "Carlos": 800
}
````

````python
steps = 0

for client in clients_list:
    steps += 1
    if client[0] == "Carlos":
        balance = client[1]
        break

print("List balance:", balance)
print("List steps:", steps)
````

````python
steps = 1  # one direct lookup

balance = clients_dict["Carlos"]

print("Dict balance:", balance)
print("Dict steps:", steps)
````

# Scope

## Variable Scope in Python

**Scope** refers to the region of a program where a variable is accessible. It determines *where a variable can be used* and *how long it exists in memory*.

Python follows the **LEGB rule**:
- **L**ocal – inside a function
- **E**nclosing – inside outer (nested) functions
- **G**lobal – defined at the top level
- **B**uilt-in – pre-defined names like `print`, `len`

---

### 1. Local Scope

Variables defined inside a function are only accessible within that function.



````python
def my_function():
    x = 10  # local variable
    print(x)

my_function()  # Output: 10
print(x)       # Error: x is not defined
````

### 2. Global Scope

Variables defined outside functions can be accessed anywhere in the file.

````python
x = 20  # global variable

def my_function():
    print(x)  # accessing global variable

my_function() 
print('_____________')
def outer():
    x = 5  # enclosing variable
    
    def inner():
        print(x)  # accesses enclosing variable
    
    inner()

outer() 
````

### 3. Enclosing Scope (Nested Functions)

Inner functions can access variables from outer functions.

````python
def outer():
    x = 5
    
    def inner():
        nonlocal x
        x = 10
    
    inner()
    print(x)

outer()  # Output: 10
````

### Common failure:
This fails because Python treats x as local, but it’s used before assignment.

````python
x = 10

def my_function():
    x = x + 5  # Error!

my_function()
````
