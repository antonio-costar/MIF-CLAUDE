# Introduction to Control Flow: `if`, `for`, and `while`

## Summary

| Statement | Main purpose | Repeats? | Typical use |
|-----------|---------------|----------|-------------|
| `if`, `elif`, `else` | Make a decision | No | Execute code only when a conditions are met |
| `for` | Iterate over items | Yes | Process each item in a collection |
| `while` | Repeat while a condition is true | Yes | Continue until a condition becomes false |

---

## The Big Picture

These statements allow us to control the behavior of a program:

- `if` asks: **"Should I do this?"**
- `for` asks: **"What should I do for each item?"**
- `while` asks: **"Should I keep doing this?"**


Control flow refers to the order in which a program executes its instructions.
By default, Python runs code from top to bottom. Control flow statements allow
us to change this behavior by making decisions or repeating instructions.

These statements are fundamental building blocks of programming because they
allow programs to respond to different situations and perform repetitive tasks
without having to write the same code multiple times.

## 1. The `if` Statement

The `if` statement is used to make a decision.

It checks whether a condition is `True`. If the condition is true, the indented
block of code underneath the `if` statement is executed.

```python
age = 40
if age >= 18:
    print("You are an adult")
```

```python
age = 10
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

## 2. The `for` Loop

A `for` loop is used to repeat a block of code for each item in a sequence or
other iterable object.

For example:

```python
for name in ["Alice", "Bob", "Charlie"]:
    print(name)
```

```python
my_list = ["Alice", "Bob", "Charlie"]
for name in my_list:
    print(name)
```

```python
for number in range(5):
    print(number)
```

## 3. The `while` Loop

A `while` loop repeatedly executes a block of code as long as a condition is
`True`.

Example:

```python
count = 0

while count < 5:
    print(count)
    count = count + 1
```
